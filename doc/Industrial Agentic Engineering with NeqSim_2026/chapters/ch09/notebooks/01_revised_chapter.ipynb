## Re-run the methane comparison

The NIST table is retained separately and matched by state, rather than row position. Agreement is reported for these states only.

```python
for row in baselines["methane_properties"]:
    for model in ("SRK", "PR"):
        fluid = make_fluid(jneqsim, row["pressure_bara"], 298.15, model, {"methane": 1.0})
        jneqsim.thermodynamicoperations.ThermodynamicOperations(fluid).TPflash()
        fluid.initProperties()
        calculated = float(fluid.getDensity("kg/m3"))
        assert abs(calculated - row[model + "_density_kg_m3"]) < 1e-6
    print(row)
```
